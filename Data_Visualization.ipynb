```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("taxis")
print(df.head())
```

```python
print("Missing values per column:", df.isnull().sum())
```

```python
numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns
for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())
print(df.head())
```

```python
categorical_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns
for col in categorical_cols:
    if not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])
print(df.isnull().sum())
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("taxis")
print(df)
```

```python
df['pickup'] = pd.to_datetime(df['pickup'])
line_df = df.sort_values('pickup').set_index('pickup')
line_df['fare'].plot(kind='line', figsize=(12, 6), title="Fare Over Time", color='blue')
plt.show()
```

```python
borough_fare = df.groupby('pickup_borough')['fare'].sum()
borough_fare.plot(kind='bar', figsize=(8, 5), title="Total Fare for Each Pickup Borough", color='skyblue', edgecolor='black', rot=0)
plt.show()
```

```python
payment_counts = df['payment'].value_counts()
payment_counts.plot(kind='pie', figsize=(7, 7), title="Distribution of Trips Based on Payment Method", autopct='%1.1f%%', startangle=140, colors=['gold', 'lightcoral', 'lightskyblue', 'lightgreen'])
plt.ylabel("")
plt.show()
```

```python
df['distance'].plot(kind='hist', bins=30, figsize=(10, 5), title="Distribution of Trip Distance", color='purple', edgecolor='black', alpha=0.7)
plt.show()
```

```python
df = sns.load_dataset("taxis")
df.boxplot(column='tip', by='pickup_borough', figsize=(10, 6), grid=False)
plt.title("Distribution of Tip Amounts for Each Pickup Borough")
plt.suptitle("")
plt.xlabel("Pickup Borough")
plt.ylabel("Tip Amount ($)")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset("taxis")
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='pickup_borough', data=df, palette='Set2')
plt.title("Number of Trips in Each Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Count of Trips")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='distance', y='fare', hue='pickup_borough', data=df, alpha=0.7)
plt.title("Relationship Between Distance and Fare")
plt.xlabel("Distance")
plt.ylabel("Fare")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
corr_matrix = df[['distance', 'fare', 'tip', 'tolls', 'total']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

```python
sns.pairplot(df[['distance', 'fare', 'tip', 'total', 'pickup_zone']].dropna(), hue='pickup_zone')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.violinplot(x='payment', y='fare', data=df, palette='Set2')
plt.title("Distribution of Fare for Each Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Fare ($)")
plt.show()
```
